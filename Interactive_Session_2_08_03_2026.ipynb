```python
import torch
import torch.nn as nn

torch.manual_seed(0)

W = nn.Linear(4, 4)
optimizer = torch.optim.SGD(W.parameters(), lr=0.1)
loss_fn = nn.MSELoss()

x = torch.randn(4, 1, 4)
y = torch.randn(4, 1, 4)

h = torch.zeros(1, 4)
total_loss = 0

for t in range(4):
    h = torch.tanh(W(h) + x[t])
    step_loss = loss_fn(h, y[t])
    total_loss = total_loss + step_loss
    print(f"step {t}:  L{t} = {step_loss.item():.4f}   grad so far: {W.weight.grad}")

print(f"\ntotal loss = {total_loss.item():.4f}")

total_loss.backward()
print(f"\ngradient after backward():\n{W.weight.grad}")

optimizer.step()
print("\nweights updated. One update for the whole sequence.")
```

```python
import torch

def propagate(w, steps=50):
    grad = 1.0
    trace = []
    for t in range(steps):
        grad = grad * w
        trace.append(grad)
    return trace

for w in [1.1, 0.9]:
    trace = propagate(w)
    print(f"\nw = {w}")
    for t in [0, 9, 19, 29, 49]:
        print(f"  after {t+1:2d} steps:  gradient = {trace[t]:.2e}")
```
